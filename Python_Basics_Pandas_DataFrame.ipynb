Question 1 :  Write a Python program to check whether a number is even or odd using conditional statements.

```python
1. Here is the Python program to check whether a number is Even or Odd using conditional statements (if-else).

Python code


num = int(input("Enter a number: "))

if num % 2 == 0:
    print("Even number")
else:
    print("Odd number")


Output

Example 1:

Enter a number: 8
Even number

Example 2:

Enter a number: 5
Odd number

2. Explanation
- input() takes a number from the user.
- int() converts the input into an integer.
- % is the modulus operator, which gives the remainder.
- if num % 2 == 0 checks whether the remainder is zero.
- If the remainder is 0, the number is even; otherwise, it is odd.
```

Question 2 : Write a Python program to count how many times a number appears in a list.

Example List

[2,3,4,2,5,2]

```python
1. Python Program


numbers = [2, 3, 4, 2, 5, 2]

num = int(input("Enter a number to count: "))

count = numbers.count(num)

print("The number", num, "appears", count, "times.")

Output

Enter a number to count: 2
The number 2 appears 3 times.

2. Explanation
- numbers is the list of numbers.
- input() takes a number from the user.
- .count(num) counts how many times that number appears in the list.
- print() displays the result.
```

Question 5 : Convert a 1D array into 2x3 matrix:

arr = np.array([1,2,3,4,5,6]).

```python
1. Python code

import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])

matrix = arr.reshape(2, 3)

print(matrix)

. Output

[[1 2 3]
 [4 5 6]]

2. Explanation

- np.array() creates a 1D array containing 6 elements.
- reshape(2, 3) converts it into a matrix with 2 rows and 3 columns.
- The total number of elements remains 6 (2 × 3 = 6).
```

Question 9 : Create a random array of 10 numbers and:

1. Find max and min.

2. Sort the array.

```python
1. Python code.

import numpy as np

# Create a random array of 10 numbers
arr = np.random.randint(1, 100, 10)

# Display the array
print("Original Array:", arr)

# Find maximum and minimum
print("Maximum Number:", np.max(arr))
print("Minimum Number:", np.min(arr))

# Sort the array
print("Sorted Array:", np.sort(arr))

2. Simple Output

Original Array: [45 12 78 34 89 23 56  9 67 30]

Maximum Number: 89
Minimum Number: 9

Sorted Array: [ 9 12 23 30 34 45 56 67 78 89]

Explanation
1. np.random.randint(1, 100, 10) – Generates 10 random numbers.
2. np.max(arr) – Finds the maximum number.
3. np.min(arr) – Finds the minimum number.
4. np.sort(arr) – Sorts numbers in ascending order.

Note: Your output may be different because the numbers are generated randomly.
```

Question 10 : Replace all even numbers with 0:

arr = np.array([1,2,3,4,5,6])

```python
1. Python code.

import numpy as np

# Create an array
arr = np.array([1, 2, 3, 4, 5, 6])

# Replace all even numbers with 0
arr[arr % 2 == 0] = 0

# Display updated array
print("Updated Array:", arr)

2. Output

Updated Array: [1 0 3 0 5 0]

Explanation
1. arr % 2 == 0 – Checks which numbers are even.
2. arr[...] = 0 – Replaces all even numbers with 0.
3. print(arr) – Displays the updated array.
Answer: [1 0 3 0 5 0]
```
